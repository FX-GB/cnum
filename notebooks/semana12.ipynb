## Atividade 1
Escreva os métodos regra de quadratura pelo ponto médio, Trapézio e Simpson dentro do arquivo $utils/algoritmos.py$.

````python
def medio(f, a, b):
    h = b - a
    return h * f((a + b) / 2)


def trapezio(f, a, b):
    h = b - a
    return h * (0.5 * f(a) + 0.5 * f(b))


def simpson(f, a, b):
    h = (b - a) / 2
    return h * ((1 / 3) * f(a) + (4 / 3) * f((a + b) / 2) + (1 / 3) * f(b))


def integral(metodo, f, a, b, n=1e-3):
    s = 0.0
    c = a
    d = a + n
    while d <= b:
        s += metodo(f, c, d)
        c = d
        d += n
    return s
````

## Atividade 2
Calcule numericamente o valor de  

$\displaystyle \int_{2}^{5} e^{4 - x^2} \, dx$  

usando os métodos **compostos** do **ponto médio**, **Trapézio** e **Simpson**.  
Obtenha os resultados utilizando, em cada quadratura, o número de pontos indicado.

**Respostas esperadas:**

| $n$ | Ponto médio | Trapézios | Simpson |
|:--:|:--:|:--:|:--:|
| 3 | 0.1056606 | 0.7503919 | 0.5005225 |
| 5 | 0.1726140 | 0.3964724 | 0.2784992 |
| 7 | 0.1973663 | 0.3062023 | 0.2393551 |
| 9 | 0.2084204 | 0.2721145 | 0.2306618 |

````python
import numpy as np
from utils.algoritmos import integral, medio, trapezio, simpson

f = lambda x: np.exp(4 - x**2)
a = 2.0
b = 5.0

print("n | Ponto médio | Trapézios | Simpson")
for n in (3, 5, 7, 9):
    subintervalos = n - 1
    r_medio = integral(medio, f, a, b, n=n)
    r_trapezio = integral(trapezio, f, a, b, n=subintervalos)

    # Simpson composto usa n subintervalos, portanto n deve ser par.
    subintervalos = n - 1
    r_simpson = integral(simpson, f, a, b, n=subintervalos)

    print(f"{n} | {r_medio:.7f} | {r_trapezio:.7f} | {r_simpson:.7f}")
````
